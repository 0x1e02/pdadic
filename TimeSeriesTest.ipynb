```python
HEIGHT = 4
DEGREE = 3
ALPHA = .1

# qtree = NAryTree(DEGREE, HEIGHT)
# qtree = QTree([9,3,3])
qtree = QTree([3,3,9])
# qtree = QTree([3,3,3,3])

func = lambda x: np.exp(-16*( x - 0.5)**2)
# func = lambda x: 1
qfunc = QTreeMap.from_function(qtree, func)

kernel = lambda x: np.abs(x)**(-ALPHA)
qoperator = QTreeOperator(qtree, kernel)

print("The eigenvalues are:")
print(qoperator.eigenvalues)
```

```python
qdecomp = qfunc.to_decomposition()

print(qdecomp
    .transform(np.zeros(qtree.height))
    .to_map()
    .values
)
print(qdecomp.mean)

# Conclusion: Transform works
```

```python
T = profile_equilibrium(qoperator, qfunc, eps=1e-3)
print(f"Profiled time until equilibrium is {T=}")

ts = T * QTreeOperator.get_geometric_ts(
    n = len(qfunc),
    r = .05
)


slices = qoperator.get_slices(qfunc, ts)

frames = np.array([
    list(s) for s in slices
])

plt.matshow(np.real(frames))
plt.show()
```
